```python
# -*- coding: utf-8 -*-
"""
Created on Sun Oct 11 21:45:33 2026

@author: 87727
"""


import random

# 2.1 create the 2 matrixes with randint
M1 = [[random.randint(0, 50) for _ in range(10)] for _ in range(5)]

M2 = [[random.randint(0, 50) for _ in range(5)] for _ in range(10)]

# 2.2 definition of the multiplication of 2 matrixes
def Matrix_multip(A, B):
    # the rows of A
    rows_A = len(A)
    # the columes of A
    cols_A = len(A[0])
    # the columes of B
    cols_B = len(B[0])

    # determine a matrix of the result , get the number of rows of A and columes of B
    result = []
    for i in range(rows_A):
        row = []
        for j in range(cols_B):
            row.append(0)
        result.append(row)

    # multiplication
    #rols of result
    for i in range(rows_A):
        # columes of the result
        for j in range(cols_B):
            # determine the result of each element 
            total = 0
            # do the math
            for k in range(cols_A):
                total = total + A[i][k] * B[k][j]
            result[i][j] = total

    return result


# result test
print("M1:")
for row in M1:
    print(row)
    
print("M2:")
for row in M2:
    print(row)

print("M1 * M2:")
answer = Matrix_multip(M1, M2)
for row in answer:
    print(row)
```

```python
# -*- coding: utf-8 -*-
"""
Created on Sun Oct 11 22:06:07 2026

@author: 87727
"""

def Pascal_triangle(k):
   
    # the first line: [1]
    line = [1]

    # If k is 0, just print [1]
    if k == 0:
        print(line)
        return

    # Build the line step by step until we reach row k
    # Each new line is built from the previous one
    for i in range(1, k + 1):
        # Create a new list that starts with 1
        new_line = [1]

        # For every pair of neighbours in the old line,
        # add them together to get the middle values
        for j in range(len(line) - 1):
            new_line.append(line[j] + line[j + 1])

        # The new line also ends with 1
        new_line.append(1)

        # The new line becomes the current line for the next round
        line = new_line

    # After the loop, 'line' is the k-th line
    print(line)


# Report the required lines
print("Pascal_triangle(100):")
Pascal_triangle(100)

print("Pascal_triangle(200):")
Pascal_triangle(200)

x = int(input("which line you would like to know in Pascal trangle:"))
Pascal_triangle(x)
```
